## Parte 1: Exploração de Ideias 💡

Antes de definir seu problema específico, vamos explorar o espaço de possibilidades. Projetos SBSE+IA podem ser classificados em quatro categorias principais, dependendo do **papel que a IA desempenha** na solução.

### Categoria A: IA como Assistente 🤖

**Conceito:** A IA (geralmente um LLM) atua como um "consultor especialista" que ajuda a formular ou refinar a busca.

**Exemplos:**

1. **LLM-Enhanced Fitness Design:** Um LLM sugere métricas de qualidade para avaliação de refatorações de código baseado na descrição do projeto.
   
2. **Prompt-Guided Operator Design:** Um LLM gera operadores de mutação customizados para um problema específico de configuração de software.
   
3. **AI-Assisted Solution Encoding:** Um LLM sugere como representar soluções de um problema complexo de arquitetura de software em cromossomos de AG.
   
4. **Semantic Search Operators:** LLM analisa código-fonte e sugere trocas semanticamente válidas para mutação em problemas de otimização de código.

### Categoria B: IA como Objeto de Otimização 🎯

**Conceito:** O alvo da otimização é um componente de IA – SBSE ajusta parâmetros, arquiteturas ou prompts para melhorar desempenho.

**Exemplos:**

1. **Hyperparameter Optimization for ML Models:** AG otimiza learning rate, batch size, número de camadas de uma rede neural para classificação de bugs.
   
2. **Automated Prompt Engineering:** Algoritmo genético evolui prompts para maximizar a qualidade de geração de testes unitários por um LLM.
   
3. **Neural Architecture Search (NAS):** SBSE busca a melhor topologia de rede neural para previsão de defeitos em código.
   
4. **LLM Configuration Optimization:** Otimizar temperatura, top-k, top-p de um LLM para geração de documentação técnica.

### Categoria C: IA como Oráculo de Avaliação 🔍

**Conceito:** A IA atua como uma função de fitness inteligente, avaliando a qualidade de soluções candidatas de forma sofisticada.

**Exemplos:**

1. **LLM-as-Judge for Code Quality:** Um LLM avalia legibilidade, manutenibilidade e aderência a padrões de código gerado por SBSE.
   
2. **AI-Based Test Case Quality Scorer:** Modelo de ML treinado avalia a efetividade de casos de teste gerados por AG.
   
3. **Semantic Equivalence Checker:** LLM verifica se refatorações propostas mantêm semântica original do código.
   
4. **Bias Detection Oracle:** Modelo de fairness avalia soluções de configuração de sistemas quanto a viés algorítmico.

### Categoria D: Abordagem Híbrida 🔄

**Conceito:** Combinação de múltiplos papéis de IA na mesma solução SBSE.

**Exemplos:**

1. **Self-Improving Test Generator:** LLM gera testes iniciais (Assistente) + AG otimiza cobertura (Busca) + Modelo ML avalia qualidade (Oráculo).
   
2. **Adaptive CI/CD Optimizer:** LLM sugere configurações iniciais + AG busca ótimos + IA prevê impacto de mudanças.
   
3. **Smart Refactoring Engine:** LLM identifica code smells + SBSE propõe refatorações + IA avalia impacto arquitetural.

Engenharia de prompts para geração de testes unitários é trabalhosa e requer expertise. Pequenas mudanças no prompt podem alterar drasticamente a qualidade dos testes gerados.

**Abordagem SBSE+IA:**
- **Representação:** Prompt estruturado em 5 componentes (contexto, instrução, exemplos, constraints, formato)
- **SBSE:** AG evolui variações de cada componente do prompt
- **IA:** GPT-4 é o alvo - para cada prompt candidato, gera testes para classes Java
- **Fitness:** Cobertura de código + número de asserções + qualidade dos edge cases

**Resultados Esperados:**
Prompts otimizados automaticamente geram testes com 85% de cobertura média, versus 60% de prompts escritos manualmente por desenvolvedores.

**Insights Técnicos:**
- Representação modular do prompt permite crossover efetivo
- Mutação inteligente: LLM sugere variações de frases
- Validação em 10 projetos Java open-source

### Exemplo 3: FairCI - IA como Oráculo

**Categoria:** IA como Oráculo

**Problema:**
Sistemas de CI/CD podem inadvertidamente favorecer certos tipos de builds ou testes (viés de priorização), degradando a experiência de subgrupos de desenvolvedores.

**Abordagem SBSE+IA:**
- **Representação:** Política de escalonamento de jobs (prioridades, alocação de recursos)
- **SBSE:** AG otimiza políticas para minimizar tempo de espera médio
- **IA:** Modelo de fairness (treinado em dados históricos) avalia se política discrimina por tipo de projeto, região ou horário
- **Fitness Multi-objetivo:** (1) Eficiência (tempo médio), (2) Fairness (score do modelo ML)

**Resultados Esperados:**
Políticas otimizadas reduzem disparidade de tempo de espera entre grupos em 40%, mantendo eficiência global a 95% do ótimo.

**Insights Técnicos:**
- Modelo de fairness treinado com XGBoost em dados de 1 ano de CI/CD
- Análise de Pareto front (trade-off eficiência vs fairness)
- Implementação real em pipeline de projeto open-source

### Exemplo 4: AdaptiveTestGen - Abordagem Híbrida

**Categoria:** Híbrido (Assistente + Oráculo)

**Problema:**
Geração automática de testes de integração para APIs REST requer (1) entender semântica dos endpoints e (2) avaliar se testes cobrem cenários críticos.

**Abordagem SBSE+IA:**
- **Representação:** Sequência de chamadas HTTP com parâmetros
- **SBSE:** AG evolui sequências que maximizam cobertura de código
- **IA como Assistente:** LLM analisa documentação da API e sugere sequências iniciais semanticamente válidas
- **IA como Oráculo:** LLM avalia se sequências testam edge cases relevantes (ex: autenticação expirada, rate limiting)
- **Fitness:** 50% cobertura tradicional + 50% score de relevância do LLM

**Resultados Esperados:**
Testes gerados cobrem 92% do código e detectam 8 bugs não encontrados por testes manuais, em avaliação com 5 APIs REST reais.

**Insights Técnicos:**
- Pipeline em 3 fases: (1) LLM gera seeds, (2) AG explora, (3) LLM filtra testes redundantes
- Redução de 60% em testes de baixa qualidade comparado a AG puro
- Implementação com DEAP + OpenAI API + instrumentação de cobertura
